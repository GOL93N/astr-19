'''Define two models to fit your data and use the curve_fit() function from scipy.optimize to retrieve the best 
fit values of your models. The first model should return y = a*exp(-b*x) for a given value of a and b. The second 
model should return y = a / (x**b) for a given value of a and b.  Print your best-fit results for both models using 
an f-string. Plot in one figure both of your best fit models on top of the data. Give labels to the data and both 
models and show them in a legend in the same figure. Save your final figure as a PDF'''

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

def model1(x, a, b):
    return a*np.exp(-b*x)

def model2(x, a, b): 
    return a / (x**b)

a = 3
b = 7

x = np.linspace(0, 2*np.pi, 100)

y = model1(x, a, b)

ammount= np.linspace(2, 0.2, 100)
y_scatter = y + np.random.normal(0, ammount)
```

```python
best_fit1, _ = curve_fit(model1, x, y_scatter)
best_fit2, _ = curve_fit(model2, x[1:], y_scatter[1:])

print(f"Model 1 Fit: a={best_fit1[0]:.2f}, b={best_fit1[1]:.2f}")
print(f"Model 2 Fit: a={best_fit2[0]:.2f}, b={best_fit2[1]:.2f}")

plt.figure()
plt.plot(x, model1(x, *best_fit1),  'r', label='Fit: Model1')
plt.plot(x, model2(x, *best_fit2), 'g', label='Fit: Model2')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.savefig('CJ2_P3.pdf')
```
